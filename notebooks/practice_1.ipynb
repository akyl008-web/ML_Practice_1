```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/titanic.csv")

df.head()
```

```python
print("Размер таблицы:", df.shape)

print("\nТипы данных:")
print(df.dtypes)

print("\nПропущенные значения:")
print(df.isnull().sum())

print("\nСтатистика:")
display(df.describe())
```

```python
sns.countplot(data=df, x="Survived")
plt.title("Выжившие и погибшие")
plt.xlabel("0 — погиб, 1 — выжил")
plt.ylabel("Количество")
plt.show()
```

```python
sns.histplot(df["Age"].dropna(), bins=20)
plt.title("Возраст пассажиров")
plt.xlabel("Возраст")
plt.ylabel("Количество")
plt.show()
```

### Наблюдения по датасету Titanic

1. **Выживаемость пассажиров:** Количество погибших пассажиров больше, чем выживших. Это видно на первом графике.

2. **Возраст пассажиров:** Большинство пассажиров, для которых указан возраст, были молодыми или людьми среднего возраста. Пожилых пассажиров было значительно меньше.

3. **Пропущенные данные:** В столбце Age есть пустые значения. Перед обучением модели машинного обучения необходимо обработать эти пропуски.
